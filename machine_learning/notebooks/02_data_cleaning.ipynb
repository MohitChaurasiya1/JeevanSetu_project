```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/diabetes.csv")
print("Original shape:", df.shape)
```

```python
# A value of 0 in these columns is not possible, so treat it as missing
cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
df[cols] = df[cols].replace(0, np.nan)
print("Missing values before filling:")
print(df[cols].isnull().sum())

# Fill missing values with the median (not affected by outliers)
df[cols] = df[cols].fillna(df[cols].median())
print("Missing values after filling:", df.isnull().sum().sum())
```

```python
raw = pd.read_csv("../data/diabetes.csv")
cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
print("Zeros in raw file:")
print((raw[cols] == 0).sum())
```

```python
print(df.shape)
print(df.columns.tolist())
print(df["Outcome"].value_counts())
```

```python
import os

# Remove duplicate rows
print("Duplicates found:", df.duplicated().sum())
df = df.drop_duplicates()
print("Final shape:", df.shape)

# Save the cleaned data in the processed folder
os.makedirs("../data/processed", exist_ok=True)
df.to_csv("../data/processed/diabetes_cleaned.csv", index=False)
print("Saved: ../data/processed/diabetes_cleaned.csv")
```
