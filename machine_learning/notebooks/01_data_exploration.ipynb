```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset (change the file name if it is different)
df = pd.read_csv("../data/diabetes.csv")

print("Shape:", df.shape)
print(df.head())
print(df.describe())
```

```python
print(df["Outcome"].value_counts())
df["Outcome"].value_counts().plot(kind="bar", title="Outcome distribution")
plt.show()
```

```python
# In columns me 0 matlab missing value
cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
print((df[cols] == 0).sum())
```

```python
df.hist(figsize=(12, 9), bins=20)
plt.tight_layout()
plt.show()
print(df.corr()["Outcome"].sort_values(ascending=False))
```
